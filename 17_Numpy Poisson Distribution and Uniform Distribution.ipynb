# Numpy Poisson Distribution

#it estimates how many times an event can happen.

#two parameter - 1. lam(number of occurance or rate) 2. size

#Generate a random 1*10 dist for the occurance 2.

```python
import numpy as np
x = np.random.poisson(lam=2, size=10)
print(x)
```

#Visualization of poisson dist

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(random.poisson(lam=2, size=1000),kde=False)
plt.show()
```

#presenting both the plots in same figure normal and poisson distribution 

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(random.normal(loc=50, scale=7, size=1000),label='Normal')
sns.histplot(random.poisson(lam=50, size=1000),label='Poisson')
plt.show()
```

# Uniform Distribution 

#Uniform distribution is a probability distribution in which all values in a specified range, or equal-width intervals within that range, have equal probability density.

#three parameter - a(lower bound - 0.0), b(upper bound - 1.0), size

```python
from numpy import random 
x = random.uniform(size=(2,3))
print(x)
```

#Visualization of uniform dist

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(random.uniform(size=1000),kde=False)
plt.show()
```

### Uniform vs Normal Distribution

| Uniform Distribution | Normal Distribution |
|---|---|
| Equal probability across equal-width intervals | More values occur near the mean |
| The histogram can be roughly flat | Bell-shaped curve |
| `np.random.uniform()` | `np.random.normal()` |
| Example: Random number selection | Example: Measurement errors |
