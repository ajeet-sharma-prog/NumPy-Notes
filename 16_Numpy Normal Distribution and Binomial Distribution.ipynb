# Normal(Gaussion) Distribution

## random.normal() method- loc(mean), scale(standard deviation), size

#we are generating a random normal distribution of size 2*3

```python
from numpy import random
import numpy as np

ajeet = random.normal(size=(2,3))

print(ajeet)
```

#here we will generate a random normal dist of size 2*3 with mean at 1 and standard deviation of 2:

```python
from numpy import random
import numpy as np

ajeet = random.normal(loc=1, scale=2, size=(2, 3))

print(ajeet)
```

#visualization of normal distribution.

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.distplot(random.normal(size=1000),hist=False)
plt.show()
```

#the curve of a normal dist is also called bell curve.

# Binomial Distribution

## binomial dist - discrete dist - binary output

#there are three parameter

#1. n = number of trials
 2. p = success of probability
 3. size = specifies the number of experiments to perform

#given 10 trial for a coin which will generate 10 data points

```python
from numpy import random
x = random.binomial(n=10, p=0.5, size=10)
print(x)
```

#Visualization of binomial dist

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(random.binomial(n=10, p=0.5, size=1000))
plt.show()
```

## Normal vs Binomial Distribution

| Normal Distribution | Binomial Distribution |
|---|---|
| Continuous data | Discrete data |
| Bell-shaped | Success/Failure trials |
| Example: Height | Example: Coin toss |
| `np.random.normal()` | `np.random.binomial()` |

#I use 500 data point for binomial and under 100 data point for normal distribution

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(random.normal(loc=50, scale=5, size=1000),label='Normal')
sns.histplot(random.binomial(n=100, p=0.5, size=1000),label='Binomial')
plt.show()
```
