# Numpy Hyperbolic Function

Hyperbolic functions are mathematical functions related to exponential functions. In NumPy, they are used to calculate hyperbolic sine, cosine, tangent, and their inverses.

Like trigonometric functions, they have similar names, but their calculations are different.

## Important Hyperbolic Functions

| Function | Purpose |
|---|---|
| `np.sinh()` | Hyperbolic sine |
| `np.cosh()` | Hyperbolic cosine |
| `np.tanh()` | Hyperbolic tangent |
| `np.arcsinh()` | Inverse hyperbolic sine |
| `np.arccosh()` | Inverse hyperbolic cosine |
| `np.arctanh()` | Inverse hyperbolic tangent |

np.sinh() — Hyperbolic Sine

```python
import numpy as np

print(np.sinh(0))
print(np.sinh(1))
print(np.sinh(2))
```

np.cosh() — Hyperbolic Cosine

```python
import numpy as np

print(np.cosh(0))
print(np.cosh(1))
print(np.cosh(2))
```

np.tanh() — Hyperbolic Tangent

```python
import numpy as np

print(np.tanh(0))
print(np.tanh(1))
print(np.tanh(2))
```

Hyperbolic Functions with Arrays

```python
import numpy as np

arr = np.array([0, 1, 2, 3])

print(np.sinh(arr))
print(np.cosh(arr))
print(np.tanh(arr))
```

Inverse Hyperbolic Functions

```python
import numpy as np

print(np.arcsinh(1))
print(np.arccosh(2))
print(np.arctanh(0.5))
```

# Function and Numpy ufuncs set Operation

NumPy set operations are used to find unique elements, common elements, and differences between arrays.

In simple words, they are used to compare the elements of two arrays and find unique or common values.

## Important Set Functions

| Function | Purpose |
|---|---|
| `np.unique()` | Removes duplicate values |
| `np.union1d()` | Returns unique elements from both arrays |
| `np.intersect1d()` | Returns common elements from both arrays |
| `np.setdiff1d()` | Returns elements in the first array that are not in the second array |
| `np.setxor1d()` | Returns unique elements that are not common to both arrays |

np.unique() — Unique elements

```python
import numpy as np

arr = np.array([10, 20, 10, 30, 20, 40])

print(np.unique(arr))
```

np.union1d() — Union

```python
arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([3, 4, 5, 6])

print(np.union1d(arr1, arr2))
```

np.intersect1d() — Intersection

```python
arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([3, 4, 5, 6])

print(np.intersect1d(arr1, arr2))
```

np.setdiff1d() — Difference

```python
print(np.setdiff1d(arr1, arr2))
print(np.setdiff1d(arr2, arr1))
```

np.setxor1d() — Symmetric difference

```python
print(np.setxor1d(arr1, arr2))
```
