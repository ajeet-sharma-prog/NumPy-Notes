# Numpy joining Array

## Here for this we will pass concatenate()

```python
import numpy as np
x = np.array([1,2,3])
x1 = np.array([4,5,6])
x2 = np.concatenate((x,x1))
print(x2)
```

#joining of 2-D array along with rows(axis = 1)

```python
import numpy as np
x = np.array([[1,2],[3,4]])
x1 = np.array([[5,6],[7,8]])
x2 = np.concatenate((x,x1), axis = 1)
print(x2)
```

#joining array usiing the stack function:

```python
import numpy as np
x = np.array([1,2,3])
x1 = np.array([4,5,6])
x2 = np.stack((x,x1),axis=1)
print(x2)
```

#stacking along with row: hstack()

```python
import numpy as np
x = np.array([1,2,3])
x1 = np.array([4,5,6])
x2 = np.hstack((x,x1))
print(x2)
```

#stacking along with column

```python
import numpy as np
x = np.array([1,2,3])
x1 = np.array([4,5,6])
x2 = np.vstack((x,x1))
print(x2)
```

#stacking along with height(depth)

```python
import numpy as np
x = np.array([1,2,3])
x1 = np.array([4,5,6])
x2 = np.dstack((x,x1))
print(x2)
```
