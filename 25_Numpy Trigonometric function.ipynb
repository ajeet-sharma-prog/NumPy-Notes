# Trigonometric Function

Trigonometric functions calculate values based on angles. They are used in mathematics, physics, graphics, and data science.

The important functions in NumPy are:

| Function | Purpose |
|----------|---------|
| `np.sin()` | Sine value |
| `np.cos()` | Cosine value |
| `np.tan()` | Tangent value |
| `np.arcsin()` | Inverse sine |
| `np.arccos()` | Inverse cosine |
| `np.arctan()` | Inverse tangent |
| `np.deg2rad()` | Converts degrees to radians |
| `np.rad2deg()` | Converts radians to degrees |

np.sin() — Sine

NumPy's trigonometric functions take angles in radians, not degrees.

```python
import numpy as np

print(np.sin(np.pi / 2))
print(np.sin(0))
print(np.sin(np.pi))
```

np.cos() — Cosine

```python
import numpy as np

print(np.cos(0))
print(np.cos(np.pi / 3))
print(np.cos(np.pi))
```

np.tan() — Tangent

```python
import numpy as np

print(np.tan(0))
print(np.tan(np.pi / 4))
```

Convert Degrees to Radians
If an angle is given in degrees, use np.deg2rad().

```python
import numpy as np

angle = np.array([0, 30, 45, 60, 90])

radians = np.deg2rad(angle)

print(np.sin(radians))
```

Convert Radian to Degrees

```python
import numpy as np

radians = np.array([0, np.pi/2, np.pi])

print(np.rad2deg(radians))
```

Inverse Trigonometric Functions
Inverse trigonometric functions are used to find angles based on trigonometric ratios.

```python
import numpy as np

print(np.arcsin(1))
print(np.arccos(1))
print(np.arctan(1))
```

```python
print(np.rad2deg(np.arcsin(1)))
print(np.rad2deg(np.arctan(1)))
```
