# Exponential distribution

#it is used for describing the time till next event that is like failure or success.

#two parameter 1. scale(inverse of rate(see lam poisson)) 2. size

#here we will drow a sample for exponential dist with 2.0 scale and 2*3 size.

```python
from numpy import random
x = random.exponential(scale=2, size=(2,3))
print(x)
```

#Visualization of Exponential dist

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(random.exponential(size = 1000), kde=False)
plt.show()
```

### Exponential vs Normal Distribution

| Exponential Distribution | Normal Distribution |
|---|---|
| Models waiting time | Models values around the mean |
| Values are `0` or greater | Values can be positive or negative |
| Right-skewed shape | Bell-shaped curve |
| `np.random.exponential()` | `np.random.normal()` |

# Chi Square distribution

#it is basically used as a basis to verify the hypothesis

#two parameter 1. df(degree of freedom), 2. size

#sample for chi squared dist with df 2 size 2*3

```python
from numpy import random
x = random.chisquare(df=2, size=(2,3))
print(x)
```

#Visualization of chi square

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(random.chisquare(df=1, size=1000), kde=False)
plt.show()
```
