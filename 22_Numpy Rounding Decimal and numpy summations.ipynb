# Numpy Rounding Decimal

NumPy rounding functions are used to round decimal numbers to the nearest integer or to a specified number of decimal places.

NumPy has 5 important rounding functions:

| Function | Purpose |
|---|---|
| `np.trunc()` | Removes the decimal part of a number. |
| `np.fix()` | Rounds towards zero. |
| `np.around()` | Rounds to the nearest value. |
| `np.floor()` | Rounds down to the nearest integer. |
| `np.ceil()` | Rounds up to the nearest integer. |

trunc()

```python
import numpy as np

arr = np.array([3.166, 4.678, -2.789])

print(np.trunc(arr))
```

fix()

```python
arr = np.array([3.8, 4.2, -2.8, -2.2])

print(np.fix(arr))
```

around()

```python
arr = np.array([3.14159, 4.67891, 2.34567])

print(np.around(arr, 2))
```

floor()

```python
arr = np.array([3.8, 4.2, -2.3])

print(np.floor(arr))
```

ceil()

```python
arr = np.array([3.2, 4.7, -2.3])

print(np.ceil(arr))
```

# Numpy Summations

NumPy summation functions are used to calculate the sum of array elements or combine the sums of multiple arrays.

NumPy has two important functions:
- np.sum() — Calculates the total sum of all elements in an array.
- np.add() — Adds corresponding elements of two arrays.

np.sum()

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

print(np.sum(arr))
```

np.add()

```python
arr1 = np.array([10, 20, 30])
arr2 = np.array([1, 2, 3])

print(np.add(arr1, arr2))
```

Use of axis
In a 2D array, axis determines the direction in which the sum is calculated.

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

print(np.sum(arr, axis=0))
print(np.sum(arr, axis=1))
```

#Cumulative sum

It calculates the running total at each step.

```python
arr = np.array([1, 2, 3, 4])

print(np.cumsum(arr))
```
