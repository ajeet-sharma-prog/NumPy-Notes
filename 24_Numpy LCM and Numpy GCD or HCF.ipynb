# Numpy LCM

LCM stands for Least Common Multiple. In Hindi, it is called Laghuttam Samapavartya.

The LCM of two or more numbers is the smallest positive number that is exactly divisible by all those numbers.

Example: Find the LCM of 4 and 6.

Multiples of 4: 4, 8, 12, 16, 20, ...

Multiples of 6: 6, 12, 18, 24, ...

The smallest common multiple of both numbers is 12.

Therefore, LCM = 12.

```python
import numpy as np

print(np.lcm(4, 6))
```

```python
x = 4
x1 = 6
x2 = np.lcm(x,x1)
print(x2)
```

#Finding the lcm in array

```python
import numpy as np

arr1 = np.array([4, 6, 8])
arr2 = np.array([6, 9, 12])

print(np.lcm(arr1, arr2))
```

LCM of Multiple Numbers
If you want to calculate the LCM of more than two numbers, you can use np.lcm.reduce().

```python
import numpy as np

arr = np.array([4, 6, 8])

print(np.lcm.reduce(arr))
```

# Numpy GCD or HCF

GCD stands for Greatest Common Divisor, and HCF stands for Highest Common Factor. Both have the same meaning.

It is the largest positive number that can completely divide all the given numbers.

Example: Find the HCF of 12 and 18.
- Factors of 12: 1, 2, 3, 4, 6, 12
- Factors of 18: 1, 2, 3, 6, 9, 18
The greatest common factor is 6.

Therefore, HCF = 6.

```python
import numpy as np

print(np.gcd(12, 18))
```

```python
import numpy as np
x = 6
x1 = 9
x2 = np.gcd(x, x1)
print(x2)
```

#Finding the GCD in array

```python
import numpy as np

arr1 = np.array([12, 20, 24])
arr2 = np.array([18, 30, 36])

print(np.gcd(arr1, arr2))
```

GCD of Multiple Numbers
If you want to calculate the GCD of more than two numbers, you can use np.gcd.reduce().

```python
import numpy as np

arr = np.array([12, 18, 24])

print(np.gcd.reduce(arr))
```
