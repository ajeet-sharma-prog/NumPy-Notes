# Numpy Product

NumPy product functions are used to calculate the product of array elements or multiply corresponding elements of arrays.

product :- use prod() function

```python
import numpy as np

arr = np.array([2, 3, 4, 5])

print(np.prod(arr))
```

```python
import numpy as np

arr = np.array([1, 2, 3, 4])
arr2 = np.array([5, 6, 7, 8])
arr3 = np.prod([arr, arr2])
print(arr3)
```

axis=0 — column-wise product

axis=1 — row-wise product

```python
import numpy as np

arr = np.array([1, 2, 3, 4])
arr2 = np.array([5, 6, 7, 8])
arr3 = np.prod([arr, arr2], axis=1)
arr4 = np.prod([arr, arr2], axis=0)
print(arr3, arr4)
```

# Cumulative product

It gives the running multiplication result at each step.

```python
arr = np.array([2, 3, 4, 5])

print(np.cumprod(arr))
```

# Numpy Differencess

NumPy difference functions are used to calculate the difference between consecutive elements of an array.

```python
import numpy as np

arr = np.array([10, 15, 25, 40])

print(np.diff(arr))
```

Use of the n Parameter
The n parameter specifies how many times the difference should be calculated.
- n=1 → Calculates the first-order difference.
- n=2 → Calculates the second-order difference.

```python
import numpy as np

arr = np.array([10, 15, 25, 40, 60])

print(np.diff(arr, n=1))
print(np.diff(arr, n=2))
```

np.diff() with a 2D Array
In a 2D array, the axis parameter determines the direction in which the difference is calculated.

```python
arr = np.array([[10, 20, 30],
                [15, 25, 40]])

print(np.diff(arr, axis=0))
print(np.diff(arr, axis=1))
```
