# Rayleigh Distribution

#it is used in signal processing 

#two parameter 1. scale(standard deviation, how flat the distribution is .. default 1.0), 2. size

#sample for RI with scale of 2.0 with size of 2*3

```python
from numpy import random
x = random.rayleigh(scale=2, size=(2,3))
print(x)
```

#Visualization of Rayleigh dist

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(random.rayleigh(size=1000), kde=False)
plt.show()
```

### Rayleigh vs Normal Distribution

| Rayleigh Distribution | Normal Distribution |
|---|---|
| Values are zero or greater | Values can also be negative |
| Usually right-skewed | Bell-shaped and symmetric |
| `np.random.rayleigh()` | `np.random.normal()` |

# Pareto Distribution

#80:20 ratio. (20% factors cause 80% outcome or result)

#two parameter 1. a(shape parameter), 2. size

#sample for pareto dist with shape 2 with size 2*3

```python
from numpy import random
x = random.pareto(a=2, size=(2,3))
print(x)
```

#Visualization of pareto dist

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(random.pareto(a=2, size=1000), kde=False)
plt.show()
```
