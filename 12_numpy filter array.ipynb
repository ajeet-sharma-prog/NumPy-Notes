# Numpy Filter Array

## Getting some element out of an existing array and creating a new array is called

#A boolean index list is a list of boolean corresponding to indexes in the array.(True and False).

#Create an array from the element on index 0 to 2:

```python
import numpy as np
x = np.array([41,42,43,44])
x1 = [True, False, True, False]
x2 = x[x1]
print(x2)
```

#now we will create a filter array

#that will return only values higher than 42.

```python
import numpy as np
X = np.array([41,42,43,44])
emptyX = []
for element in X:
    if element > 42:
        emptyX.append(True)
    else:
        emptyX.append(False)
x2 = X[emptyX]
print(emptyX)
print(x2)
```

#Create a filter array that will return only even element from the original array.

```python
import numpy as np
x = np.array([1,2,3,4,5,6,7])
amanempty = []
for i in x:
    if i % 2 == 0:
        amanempty.append(True)
    else:
        amanempty.append(False)
x2 = x[amanempty]
print(amanempty)
print(x2)
```

#yes, you can also create filtter directly from array.

#that will return only values higher than 42.

```python
import numpy as np
x = np.array([41,42,43,44])
amanempty = x > 42
x2 = x[amanempty]
print(amanempty)
print(x2)
```
