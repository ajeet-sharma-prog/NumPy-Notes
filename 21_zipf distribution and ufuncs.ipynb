#Sample for zipf dist with a as 2 with size 2*3

```python
from numpy import random 
x = random.zipf(a=2, size=(2,3))
print(x)
```

#Visualization of zipf dist

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

x = random.zipf(a=2, size=1000)

sns.histplot(x, kde=False)
plt.show()
```

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

x = random.zipf(a=2, size=1000)

sns.histplot(x[x<10], kde=False)
plt.show()
```

# ufuncs

#ufuncs also takes additional arguments like, where dtype and out.

#vectorization - converting the iterative statements into a vector based statement

example without ufuncs, here we will use python in built zip().

```python
x = [1, 2, 3, 4]
y = [4, 5, 6, 7]
z = []
for i, j in zip(x,y):
    z.append(i+j)
print(z)
```

with ufuncs, we will now use add() function.

```python
import numpy as np

x = [1, 2, 3, 4]
y = [4, 5, 6, 7]

z = np.add(x, y)
print(z)
```

#to create your own ufuncs you have to define a function like you do in normal function in python then you add it to the numpy function with from pyfunc() method

#arguments of frompy func(): function, inputs, outputs

#create your own ufunc for addition:

```python
import numpy as np
def myadd(x,y):
    return x+y
myadd = np.frompyfunc(myadd, 2, 1)
print(myadd([1,2,3,4],[5,6,7,8]))
```

#power() function raises the value from the 1st array to the power of the values of the 2nd array and return the results in a new array.

```python
import numpy as np
x = np.array([10, 20, 30, 40, 50, 60])
x1 = np.array([3, 5, 6, 8, 2, 33])
x3 = np.power(x, x1)
print(x3)
```

#Remainder- mod() and reminder() function return the remainder of the 1st array corrosponding to the 2nd array and result in the array

. mod() using

```python
import numpy as np
x = np.array([10, 20, 30, 40, 50, 60])
x1 = np.array([3, 7, 4, 8, 2, 33])
x3 = np.mod(x, x1)
print(x3)
```

. remainder() using

```python
import numpy as np
x = np.array([10, 20, 30, 40, 50, 60])
x1 = np.array([3, 7, 4, 8, 2, 33])
x3 = np.remainder(x, x1)
print(x3)
```

#Quotient and mod(remainder)

the divmod() function return both the quotient and mod

```python
import numpy as np
x = np.array([10, 20, 30, 40, 50, 60])
x1 = np.array([3, 7, 4, 8, 2, 33])
x3 = np.divmod(x, x1)
print(x3)
```
