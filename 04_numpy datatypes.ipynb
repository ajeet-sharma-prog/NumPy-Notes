data type in python - String, integer, float, boolean, complex

data type in numpy 

1. i for integer
 
2. b for boolean

3. u for unsigned integer
 
4. f for float
 
5. c for complex float
 
6. m for timedelta
 
7. M for datetime
 
8. O for object

9. S for string

10. U for unicode string

11. V for memory

#checking the data type of numpy array dtype

```python
import numpy as np
ajeet = np.array([1,2,3,4])
print(ajeet.dtype)
```

#checking the data type of numpy array string

```python
import numpy as np
ajeet = np.array(['apple', 'banana', 'cherry'])
print(ajeet.dtype)
```

#creating numpy with a defined data type

```python
import numpy as np
ajeet = np.array([1,2,3,4], dtype = 'S')
print(ajeet)
print(ajeet.dtype)
```

#now we will create an array with data type of 4 byte int

```python
import numpy as np
ajeet = np.array([1,2,3,4], dtype='i4')
print(ajeet)
print(ajeet.dtype)
```

#if a type is given in which the element can not be casted then numpy will raise error. what if a value cannot be converted?

```python
import numpy as np
ajeet = np.array(['a', '2','3'],dtype='i')
print(ajeet.dtype)
```

#concerting data type in existing array - astype()

```python
import numpy as np
ajeet = np.array([1.1, 2.0, 3.1])
ajeet1 = ajeet.astype('i')
print(ajeet1)
print(ajeet1.dtype)
```

#converting data type from integer to boolean

```python
import numpy as np
ajeet = np.array([1,0,3])
ajeet1 = ajeet.astype(bool)
print(ajeet1)
print(ajeet1.dtype)
```
